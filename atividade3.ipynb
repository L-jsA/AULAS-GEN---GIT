# Carregamento de dados

```python
df_carbon_offsets = pd.read_csv('/content/carbon_offsets.csv')
df_customers_data= pd.read_csv('/content/customers_data.csv')
df_rentals_data = pd.read_json('/content/rentals_data.json')
df_vehicles_data = pd.read_json('/content/vehicles_data.json')
```

1. customers --> age --> valor ausente
2. rentals --> daily_rate com valor < 0
3. vehicles --> co2_g_per_km dados ausente
4. rentals --> daily_rate está como string
5. rentals --> data inválida
6. rentals --> data com formatação diferente

```python
# customers --> age --> valor ausente

# Idade ausente
media_idade = df_customers_data['age'].mean()

df_customers_data['age'] = df_customers_data['age'].fillna(media_idade)

display(df_customers_data)
```

```python
# rentals --> daily_rate com valor < 0
df_rentals_data.loc[df_rentals_data['daily_rate']<=0, 'daily_rate'] = pd.NA
display(df_rentals_data)
```

```python
# vehicles --> co2_g_per_km dados ausente

df_vehicles_data['co2_g_per_km'] = df_vehicles_data['co2_g_per_km'].fillna(120)

display(df_vehicles_data)
```

```python
# rentals --> daily_rate está como string
df_rentals_data['daily_rate'] = pd.to_numeric(df_rentals_data['daily_rate'], errors='coerce')
display(df_rentals_data)
```

```python
# rentals --> data inválida
df_rentals_data['start_date'] = pd.to_datetime(df_rentals_data['start_date'], errors = 'coerce')
df_rentals_data['end_date'] = pd.to_datetime(df_rentals_data['end_date'], errors = 'coerce')
print('quantidade de dados truncados') # Exibe uma mensagem no console
print(df_rentals_data['start_date'].isna().sum()) # Conta e mostra quantas datas ficaram inválidas/nulas após a conversão.

display(df_rentals_data)
```

```python
# rentals --> data com formatação diferente
```

```python
# Criar backup de dados
df_carbon_clean = df_carbon_offsets
df_rentals_clean = df_rentals_data
df_customer_clean = df_customers_data
df_vehicles_clean = df_vehicles_data
```

```python
# Calcule a duração de cada locação (em dias)
df_rentals_clean['dias_locação'] = (
    df_rentals_clean['end_date'] - df_rentals_clean['start_date']
).dt.days

```

```python
# Calcule a receita total por locação
# Valor da diária = daily_rate

# dias de locação * valor da diária
df_rentals_clean['receita_total'] = df_rentals_clean['dias_locação'] * df_rentals_clean['daily_rate']
display(df_rentals_clean)
```

```python
# Faça o merge com a base de veículos

df_consolidado = df_rentals_clean.merge(df_vehicles_clean, on='vehicle_id')
df_conslidado = df_consolidado.merge(df_customer_clean, on='customer_id')

display(df_consolidado)
```

```python
# Calcule a emissão de CO2 por locação
df_consolidado['co2_emitido'] = df_conslidado['co2_g_per_km'] * df_conslidado['distance_km']
display(df_consolidado)
```

```python
# Extraia ano e o mês da data de inicio de locação
# para extrair informações da data temos:
# .dt.days - dia
# .dt.month - mês
# .dt.year - ano
# .dt.weekday - dia do mês

df_consolidado['mes'] = df_consolidado['start_date'].dt.month
df_consolidado['ano'] = df_consolidado['start_date'].dt.year
df_consolidado['dia_da_semana'] = df_consolidado['start_date'].dt.weekday

display(df_consolidado)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Diagrama de dispersão
# X é tempo
# Y são vendas
x = [10,20,30,40,50]
y = [15,25,20,22,30]

# Criar visualização
plt.scatter(x,y, color = 'blue', label = 'Gráfico no tempo') # scatter significa +/- dispersão
plt.title('Vendas no tempo', fontsize = 14)
plt.xlabel('Tempo', fontsize = 11)
plt.ylabel('Vendas', fontsize = 11)
plt.plot(x,y, color = 'red', label = 'linhas', linestyle = '-')
plt.legend()
plt.grid(True)
plt.show()

```

```python
# Gerar gráfico de barra

categoria = ['A','B','C','D', 'E'] # string
valores = [10,20,30,45,8] # int

# Plotar o gráfico
plt.bar(categoria, valores, color = 'skyblue', edgecolor = 'black')
plt.title('Gráfico de barras por categorias', fontsize = 14)
plt.xlabel('Categorias', fontsize = 11)
plt.ylabel('Valores', fontsize = 11)
plt.grid(axis = 'y', linestyle = '--', alpha = 0.7)
```

```python
# Gráfico de linhas
plt.plot(x,y, color = 'blue', label = 'vendas mensais', linestyle = '-')
plt.title('Acompanhamento das vendas', fontsize = 14)
plt.xlabel('Meses', fontsize = 11)
plt.ylabel('Vendas (milhões)', fontsize = 11)
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Montar bloxplot
vendas_mensais = [300,300,400,450,450,500]

# plt.figure a gente pode usar para box plot e também gráfico de pizza
# para gráficos de setores usa "embaixo" plt.pie
# para boxplot usa o plt.boxplot

plt.figure(figsize=(8,6))
plt.boxplot(vendas_mensais, patch_artist=True,
            boxprops=dict(facecolor='skyblue', color='black'),medianprops=dict(color='red'))
```

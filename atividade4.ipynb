Se um código está gerando resultados errados de quem é a culpa? Do código ou da falta de feedback?


*   Dados errados --> decições erradas
*   Códigos sem revisão --> retrabalhos
*   Feedback cedo demaias --> menos custo





```python
# Dados ruins para decisões ruins

import pandas as pd

dados = {
    'mes': ['jan', 'fev','Mar', 'abr'],
    'vendas': [1000,2000,5000, None],
    'regiao': ['Sul','Sul','Sul', 'Sul']
}

df_dados = pd.DataFrame(dados)

# CRIANDO a coluna vendas_none
df_dados['vendas_none'] = df_dados['vendas']

media_vendas = df_dados['vendas'].mean()
print(media_vendas, 'correto')

media_none = df_dados['vendas_none'].mean()
print(media_none, 'none')

print(df_dados['vendas'][3])
```

```python
# solução para o caso none

# para arrumar um campo com "erro de informação" usamos o comando - fillna
# cria a coluna vendas_none
df_dados['vendas_none'] = df_dados['vendas']

# corrige os valores None usando fillna
df_dados['vendas_corrigido'] = df_dados['vendas_none'].fillna(0)

display(df_dados)
```

```python
# Ruim
import matplotlib.pyplot as plt

# plt.bar(x,y)
plt.bar(df_dados['mes'],df_dados['vendas_corrigido'])
plt.show()
```

```python
plt.bar(df_dados['mes'],df_dados['vendas_corrigido'])

plt.title('vendas mensais em R$')

plt.xlabel('Mes')

plt.ylabel('Valores em R$')

plt.grid()

plt.show()
```

```python
# métricas erradas

dados2 = {
    'mes': ['jan', 'fev','mar','abr'],
    'vendas':[1000,1200,1100,1300]
}

df_dados2 = pd.DataFrame(dados2)

crescimento_vendas = df_dados2['vendas'].pct_change().mean()
print(crescimento_vendas)

# arrumando o valor
print(crescimento_vendas*100, '%')
#arrumando casas decimais
# f'{variavel:. QUANTIDADE DE CASAS DECIMAIS f'}
print(f'{crescimento_vendas*100:.2f}%')
# ou
print(f"{crescimento_vendas:.2%}")
```

```python
media_2 = df_dados2['vendas'].rolling(2).mean()
media_3 = df_dados2['vendas'].rolling(3).mean()

plt.plot(df_dados2['mes'], df_dados2['vendas'], label='Vendas')
plt.plot(df_dados2['mes'], media_2, label='Media 2 meses')
plt.plot(df_dados2['mes'], media_3, label='Media 3 meses')
plt.legend()
plt.show
```

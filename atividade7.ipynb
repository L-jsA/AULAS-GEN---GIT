```python
# supervisionado - tenho rótulos de dados para aprender
# o sistema aprende via categorias

# nao supervisionado - identifica padrões em dados nao rotulados

# reforço - aprende via interações com o sistema

# para um sistema aprender ele precisa:
# treino, teste e sobreajuste

# dados de treino - são modelos de resposta que queremos copiar
# teste é a validação do treino
# sobreajuste é o reforço do treino
```

```python
# A biblioteca de machine learning
# Sklearn

# Comparativo de horas estudadas com a nota da prova

import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split

horas_estudadas = np.random.rand(50,1) * 10


notas = 2* horas_estudadas + 1 + np.random.rand(50,1) * 2

# os dados tem que ser divididos entre treinop e teste
# 80% treino e 20% teste

horas_estudadas_treino, horas_estudadas_teste, notas_treino, notas_teste = train_test_split(horas_estudadas, notas, test_size=0.5, random_state=42)

plt.scatter(horas_estudadas_treino, notas_treino, color='blue')
plt.scatter(horas_estudadas_teste, notas_teste, color='red')
plt.xlabel('horas estudadas')
plt.ylabel('nota')
plt.show()
```

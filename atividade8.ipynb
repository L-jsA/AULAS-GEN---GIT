```python
# 1 - coleta de dados: receber as informações que podem vir de diversas fontes
# 2 - limpeza dos dados: remover informações não necessáriass (dados em branco, vazios, erros...)
# 3 - engenharia de atributos: com o que temos a fazer?
# 4 - treinar modelo
# 5 - testar o modelo

# sempre que trabalhamos com dados e com ml precisamos:
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
# bibliotecas genericas para dados

# vamos usar a biblioteca skleran para treinar e testar o modelo de previsão das notas
# existem métricas padrões que podemos utilizar:
# erro quadratico médio
# erro médio absoluto
# score

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

df_notas = pd.read_excel("/content/Copy of BIG-DATA-05 - Predicting School Exam Scores.xlsx")
df_notas.head()
```

```python
# compreender os dados
df_notas.info()
df_notas.describe()
```

```python
# na limpeza você pode utilizar da média ou da exclusão dos dados
df_notas.isnull().sum()

# colocando a média
# tratando pelo tipo do dado
df_notas.fillna(df_notas.mean(numeric_only=True), inplace=True)

display(df_notas)
```

```python
# corrigir dados invalidos ou errados
df_notas['Attendance Rate (%)'] = df_notas['Attendance Rate (%)'].clip(upper=100)
display(df_notas)
```

```python
# engenharia de atributos
# o que temos de relevante?

# feature
caracteristicas = ["Study Hours", "Attendance Rate (%)"]
caracteristicas_notas = df_notas[caracteristicas]

# target
alvo = 'Exam Score'
alvo_notas = df_notas[alvo]

# o campo exam exore do excel ja tem resposta
# o campo Final exam score que criarmos vai prever as respostas

# em geral para os trabalhos divimos os dados para treino e teste

caracteristicas_notas_train, caracteristicas_notas_test, alvo_notas_train, alvo_notas_test = train_test_split(
    caracteristicas_notas, alvo_notas, test_size=0.2, random_state=42
)
```

```python
# treinar o modelo
modelo = LinearRegression()
modelo.fit(caracteristicas_notas_train, alvo_notas_train)
```

```python
# tentar fazer uma previsão de dados
alvo_notas_previsao = modelo.predict(caracteristicas_notas_test)
print(alvo_notas_previsao)
```

```python
# inserir casos para o sistema
# criar uma comparaçào

comparacao = pd.DataFrame({
    "Notal real": alvo_notas_test.values,
    "Previsao de nota": alvo_notas_previsao
})
comparacao.head()

novo_aluno = pd.DataFrame({
    "Study Hours": [5],
    "Attendance Rate (%)": [70]
})

nota_previsao = modelo.predict(novo_aluno)
print(f'{nota_previsao}')
```
